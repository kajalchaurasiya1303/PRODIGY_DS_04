**Importing Libraries and Displaying Data**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
data = pd.read_csv("twitter_validation.csv", header=None)

# Give names to the columns
data.columns = ["Tweet_ID", "Topic", "Sentiment", "Text"]

# Display first 5 records
print("First 5 Records:")
print(data.head())
```

**Data Cleaning**

```python
# ---------------- DATA CLEANING ----------------

# Check missing values
print("\nMissing Values:")
print(data.isnull().sum())

# Remove rows with missing values
data = data.dropna()

# Remove duplicate tweets
data = data.drop_duplicates()

print("\nDataset after Cleaning:")
print(data.shape)
```

**Exploratory Data Analysis (EDA)**

```python
# ---------------- EDA ----------------

# Display sentiment counts
print("\nSentiment Counts:")
print(data["Sentiment"].value_counts())


# 1. Overall Sentiment Distribution

plt.figure(figsize=(7, 5))

data["Sentiment"].value_counts().plot(kind="bar")

plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")
plt.title("Overall Sentiment Distribution")
plt.xticks(rotation=0)

plt.show()
```

**Sentiment By Topics**

```python
# 2. Sentiment by Topic

sentiment_topic = pd.crosstab(data["Topic"], data["Sentiment"])

sentiment_topic.plot(kind="bar", figsize=(12, 6))

plt.xlabel("Topic / Brand")
plt.ylabel("Number of Tweets")
plt.title("Sentiment by Topic / Brand")
plt.xticks(rotation=45)

plt.legend(title="Sentiment")
plt.tight_layout()

plt.show()
```

**No. of Tweets of Top10 Brands**

```python
# 3. Top Topics / Brands

topic_count = data["Topic"].value_counts().head(10)

plt.figure(figsize=(10, 5))

topic_count.plot(kind="bar")

plt.xlabel("Topic / Brand")
plt.ylabel("Number of Tweets")
plt.title("Top 10 Topics / Brands")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
